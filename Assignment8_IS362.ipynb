```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

## getting and loading data from uci repo link
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/auto-mpg/auto-mpg.data"
cols = ['mpg', 'cylinders', 'displacement', 'horsepower', 'weight',
        'acceleration', 'model_year', 'origin', 'car_name']
df = pd.read_csv(url, delim_whitespace=True, names=cols)

## fixing horesepower data and making it a number
df['horsepower'] = df['horsepower'].replace('?', np.nan)
df['horsepower'] = pd.to_numeric(df['horsepower'])

## changing origin numbers to places
df['origin'] = df['origin'].map({1: 'USA', 2: 'Asia', 3: 'Europe'})
```

```python
## making basic bar chart for showing cylinder distribution
df['cylinders'].value_counts().plot(kind='bar')
plt.title('Cylinders Distribution')
plt.xlabel('Cylinders')
plt.ylabel('Count')
plt.show()
```

```python
## making basic scatterplot for horsepower vs weight
plt.scatter(df['weight'], df['horsepower'])
plt.title('Horsepower vs Weight')
plt.xlabel('Weight')
plt.ylabel('Horsepower')
plt.show()
```

```python
## questions : displaying mgp distribution
print("Question : How does fuel efficiency/mpg compare across different regions?")

## answer
sns.boxplot(data=df,  x='origin', y='mpg')
plt.title('MPG by Origin')
plt.show()
```
